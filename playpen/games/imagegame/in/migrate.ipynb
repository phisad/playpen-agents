```python
import json

for v in ["0.9", "1.0", "1.6"]:
    fn = f"instances_v{v}.json"
    with open(fn, encoding="utf-8") as f:
        data = json.load(f)
        
    def fix_target(game):
        target = game["target_grid"]
        if not target.startswith("\n"):
            target = "\n" + target
        if not target.endswith("\n"):
            target = target + "\n"
        return target
    
    for experiment in data["experiments"]:
        for game in experiment["game_instances"]:
            game["target_grid"] = fix_target(game)
            
    with open(fn, encoding="utf-8", mode="w") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)
```
